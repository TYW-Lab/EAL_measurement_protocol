# Imports


```python
import re
from pathlib import Path
import xml.etree.ElementTree as ET

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import tifffile as tiff
```

# Utility Functions


```python
def extract_eom(folder_name):
    """Return the EOM percentage encoded as 'P=<value>' in a folder name."""
    match = re.search(r"P=(\d+)", str(folder_name))
    return int(match.group(1)) if match else None


def find_first_tif(folder):
    """Return the first TIFF path in a folder, sorted for reproducibility."""
    tif_paths = sorted(Path(folder).glob("*.tif"))
    return tif_paths[0] if tif_paths else None


def load_tif_stack(folder):
    """Load the image stack stored in a folder, or return None if no TIFF is found."""
    tif_path = find_first_tif(folder)
    return tiff.imread(tif_path) if tif_path is not None else None


def load_z_positions(folder):
    """Read PositionZ values from the companion OME metadata file.

    Returns
    -------
    np.ndarray or None
        Z positions sorted by the zero-based TheZ index. None is returned when
        no companion OME file is available.
    """
    companion_files = sorted(Path(folder).glob("*.companion.ome"))
    if not companion_files:
        return None

    tree = ET.parse(companion_files[0])
    z_values = []
    for elem in tree.getroot().iter():
        if "TheZ" in elem.attrib and "PositionZ" in elem.attrib:
            z_index = int(elem.attrib["TheZ"]) - 1
            z_position = float(elem.attrib["PositionZ"])
            z_values.append((z_index, z_position))

    z_values.sort(key=lambda item: item[0])
    return np.array([position for _, position in z_values])


def compute_background_stack(bg_dir):
    """Average all TIFF images in the background folder to estimate background."""
    images = [tiff.imread(path) for path in sorted(Path(bg_dir).glob("*.tif"))]
    if not images:
        raise FileNotFoundError(f"No TIFF files found in background folder: {bg_dir}")
    return np.mean(images, axis=0)


def top_percent_mean(image, percent):
    """Return the mean of the brightest `percent` pixels in an image."""
    flat_image = image.ravel()
    threshold = np.percentile(flat_image, 100 - percent)
    return np.mean(flat_image[flat_image >= threshold])


def collect_eom_folders(root_dir):
    """Map each EOM percentage to its acquisition folder."""
    eom_folders = {}
    for folder in sorted(Path(root_dir).iterdir()):
        if not folder.is_dir():
            continue
        eom = extract_eom(folder.name)
        if eom is not None:
            eom_folders[eom] = folder
    return eom_folders


def load_power_lookup(excel_path):
    """Load the EOM-to-power lookup table from the first two spreadsheet columns."""
    lookup_df = pd.read_excel(excel_path)
    return dict(zip(lookup_df.iloc[:, 0], lookup_df.iloc[:, 1]))


def subtract_background(stack, background_stack, clip_negative=True):
    """Subtract background from a stack and optionally clip negative pixels to zero."""
    corrected = stack.astype(float) - background_stack
    if clip_negative:
        corrected[corrected < 0] = 0
    return corrected


def build_depth_signal_data(raw_stacks, background_stack, eom_to_power, top_percent, clip_negative=True):
    """Calculate fluorescence signal at each depth for each EOM setting.

    The signal definition matches the original notebook: mean intensity of the
    brightest `top_percent` pixels after background subtraction.
    """
    depth_data = {}
    for eom, stack in raw_stacks.items():
        if eom not in eom_to_power:
            continue

        corrected_stack = subtract_background(stack, background_stack, clip_negative)
        power = eom_to_power[eom]

        for z_index, image in enumerate(corrected_stack):
            depth_data.setdefault(z_index, {"power": [], "eom": [], "signal": []})
            depth_data[z_index]["power"].append(power)
            depth_data[z_index]["eom"].append(eom)
            depth_data[z_index]["signal"].append(top_percent_mean(image, top_percent))
    return depth_data


def sorted_positive_depth_arrays(depth_record):
    """Return power, EOM, and signal arrays filtered to positive values and sorted by power."""
    powers = np.array(depth_record["power"])
    eoms = np.array(depth_record["eom"])
    signals = np.array(depth_record["signal"])

    valid = (powers > 0) & (signals > 0)
    powers, eoms, signals = powers[valid], eoms[valid], signals[valid]

    order = np.argsort(powers)
    return powers[order], eoms[order], signals[order]


def fit_loglog_slope(powers, signals):
    """Fit log(signal) = slope * log(power) + intercept."""
    return np.polyfit(np.log(powers), np.log(signals), 1)


def select_best_two_photon_window(
    powers,
    eoms,
    signals,
    target_slope=2.0,
    min_points=3,
    max_points=None,
):
    """Find the contiguous power window with slope closest to the 2PE target.

    The default `max_points=len(powers)-1` preserves the original behavior of
    considering all contiguous windows except the full power range.
    """
    n_points = len(powers)
    if n_points < min_points:
        return None, []

    if max_points is None:
        max_points = n_points - 1

    candidates = []
    for start in range(n_points):
        for end in range(start + min_points, n_points + 1):
            if (end - start) > max_points:
                continue

            slope, intercept = fit_loglog_slope(powers[start:end], signals[start:end])
            candidates.append(
                {
                    "start": start,
                    "end": end,
                    "slope": slope,
                    "intercept": intercept,
                    "eoms": eoms[start:end],
                    "powers": powers[start:end],
                    "signals": signals[start:end],
                }
            )

    if not candidates:
        return None, []

    slopes = np.array([candidate["slope"] for candidate in candidates])
    best_index = np.argmin(np.abs(slopes - target_slope))
    return candidates[best_index], candidates


def neighboring_window_slope(powers, eoms, signals, start, end):
    """Fit an adjacent comparison window and return its slope and EOM values."""
    slope, _ = fit_loglog_slope(powers[start:end], signals[start:end])
    return slope, eoms[start:end]


def coefficient_of_determination(y_true, y_pred):
    """Return R² for a fitted model."""
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    return 1 - ss_res / ss_tot
```

# Load Data


```python
# --- Configuration: file paths ---
ROOT_DIR = Path("/Users/zipeiwu/Desktop/BU/lab/projects/imaging/Multiphoton-attenuation-length-measurement")
BACKGROUND_DIR = ROOT_DIR / "bg" # Path to background image folder
LOOKUP_TABLE_PATH = ROOT_DIR / "lookuptable.xlsx" # Path to EOM-to-power lookup table (Excel file)
```

```python
# Load acquisition folders, raw image stacks, z metadata, and EOM-power calibration.
eom_folder_map = collect_eom_folders(ROOT_DIR)
eoms_sorted = sorted(eom_folder_map)
reference_eom = max(eoms_sorted)

raw_stacks = {
    eom: load_tif_stack(eom_folder_map[eom]).astype(float)
    for eom in eoms_sorted
}

reference_stack = raw_stacks[reference_eom]
z_positions = load_z_positions(eom_folder_map[reference_eom])
background_stack = compute_background_stack(BACKGROUND_DIR)
eom_to_power = load_power_lookup(LOOKUP_TABLE_PATH)
```

# Background Subtraction


```python
# --- Configuration: signal extraction and background subtraction ---
TOP_PERCENT = 1 # Percentage of brightest pixels to define the signal at each depth
CLIP_NEGATIVE_PIXELS = True
```

```python
# Build the per-depth data table used by downstream fitting.
# Each depth stores matched lists of power, EOM, and fluorescence signal.
all_data = build_depth_signal_data(
    raw_stacks=raw_stacks,
    background_stack=background_stack,
    eom_to_power=eom_to_power,
    top_percent=TOP_PERCENT,
    clip_negative=CLIP_NEGATIVE_PIXELS,
)
```

# Determination of the Power Range for Unsaturated 2PE for Each Depth


```python
# --- Configuration: unsaturated 2PE power-window search ---
TARGET_TWO_PHOTON_SLOPE = 2.0
MIN_FIT_POINTS = 3 # Minimum number of points in a candidate power window to be considered for fitting
MAX_FIT_POINTS = None  # None preserves the original max window length: N - 1.
PLOT_POWER_LAW_FITS = True
POWER_FIT_FIGSIZE = (5, 4)
```

```python
# For each depth, fit all contiguous power windows and keep the slope nearest 2.
# A log-log slope near 2 indicates an unsaturated two-photon excitation regime.
results_table = []
fit_summaries = {}

for z_index in sorted(all_data):
    powers, eoms, signals = sorted_positive_depth_arrays(all_data[z_index])
    max_fit_points = len(powers) - 1 if MAX_FIT_POINTS is None else MAX_FIT_POINTS
    best_window, _ = select_best_two_photon_window(
        powers,
        eoms,
        signals,
        target_slope=TARGET_TWO_PHOTON_SLOPE,
        min_points=MIN_FIT_POINTS,
        max_points=max_fit_points,
    )

    if best_window is None:
        continue

    depth_value = z_positions[z_index] if z_positions is not None else z_index
    best_eom = np.max(best_window["eoms"])
    best_power = np.max(best_window["powers"])

    fit_summaries[z_index] = {
        "depth": depth_value,
        "powers": powers,
        "eoms": eoms,
        "signals": signals,
        "best": best_window,
    }
    results_table.append([depth_value, best_window["slope"], best_eom, best_power])
```

```python
# Report and plot the selected 2PE window for each depth.
for summary in fit_summaries.values():
    best = summary["best"]
    powers = summary["powers"]
    eoms = summary["eoms"]
    signals = summary["signals"]
    depth_value = summary["depth"]

    left_window = (best["start"] - 1, best["end"]) if best["start"] > 0 else None
    right_window = (best["start"], best["end"] + 1) if best["end"] < len(powers) else None

    print(f"\nDepth = {depth_value:.2f} µm")
    print(f"BEST : k={best['slope']:.3f}, EOMs={best['eoms']}")

    if left_window:
        left_slope, left_eoms = neighboring_window_slope(powers, eoms, signals, *left_window)
        print(f"LEFT : k={left_slope:.3f}, EOMs={left_eoms}")

    if right_window:
        right_slope, right_eoms = neighboring_window_slope(powers, eoms, signals, *right_window)
        print(f"RIGHT: k={right_slope:.3f}, EOMs={right_eoms}")

    if not PLOT_POWER_LAW_FITS:
        continue

    log_power = np.log(best["powers"])
    log_signal = np.log(best["signals"])
    x_fit = np.linspace(log_power.min(), log_power.max(), 100)
    y_fit = best["slope"] * x_fit + best["intercept"]

    plt.figure(figsize=POWER_FIT_FIGSIZE)
    plt.scatter(log_power, log_signal, label="Data")
    plt.plot(
        x_fit,
        y_fit,
        label=f"Fit: y={best['slope']:.2f}x + {best['intercept']:.2f}",
        linestyle="--",
    )
    plt.title(f"Depth={depth_value:.2f} µm\nEOMs={best['eoms']}")
    plt.xlabel("log(Power)")
    plt.ylabel("log(Signal)")
    plt.legend()
    plt.tight_layout()
    plt.show()

print("Analysis done.")
```

# Selection of the Depth Range for EAL Calculation


```python
# --- Configuration: result export, slope plot, and EAL depth selection ---
DEPTH_ANALYSIS_PATH = ROOT_DIR / "depth_analysis_topPercent.xlsx"
DEPTH_SLOPE_PLOT_PATH = ROOT_DIR / "depth_vs_slope.png"
PLOT_DEPTH_SLOPE = True
DEPTH_SLOPE_FIGSIZE = (6, 4)
```

```python
# Save the best unsaturated 2PE power range per depth, then plot slope vs depth.
df_out = pd.DataFrame(results_table, columns=["Depth", "Slope", "Best EOM", "Power"])
df_out.to_excel(DEPTH_ANALYSIS_PATH, index=False)

print("\n===== Final Table =====")
print(df_out)

if PLOT_DEPTH_SLOPE:
    depths = df_out["Depth"].values
    best_slopes = df_out["Slope"].values

    plt.figure(figsize=DEPTH_SLOPE_FIGSIZE)
    plt.plot(depths, best_slopes, marker="o")
    plt.axhline(TARGET_TWO_PHOTON_SLOPE, linestyle="--", label="Slope = 2")
    plt.xlabel("Depth (µm)")
    plt.ylabel("Best Slope")
    plt.title("Best Slope vs Depth")
    plt.legend()
    plt.tight_layout()
    plt.savefig(DEPTH_SLOPE_PLOT_PATH, dpi=300)
    plt.show()
```

# EAL Estimation


```python
# --- Configuration: EAL plot export ---
EAL_PLOT_PATH = ROOT_DIR / "EAL.png"
EAL_FIGSIZE = (6, 4)

# These zero-based stack indices are used in the EAL calculation below.
Z_MIN_IDX = 1
Z_MAX_IDX = 11
DEPTH_STEP_UM = 10
```

```python
# Estimate EAL from log(S / P²) vs depth.
# For each selected depth, the original workflow uses the power point just below
# the best unsaturated upper EOM to stay inside the 2PE regime.
z_values_um = []
log_signal_over_power_squared = []

for z_index in range(Z_MIN_IDX, Z_MAX_IDX + 1):
    row = df_out.iloc[z_index]
    best_eom = row["Best EOM"]

    eoms = np.array(all_data[z_index]["eom"])
    powers = np.array(all_data[z_index]["power"])
    signals = np.array(all_data[z_index]["signal"])

    order = np.argsort(powers)
    eoms, powers, signals = eoms[order], powers[order], signals[order]

    matching_indices = np.where(eoms == best_eom)[0]
    if len(matching_indices) == 0 or matching_indices[0] == 0:
        continue

    use_index = matching_indices[0] - 1
    power = powers[use_index]
    signal = signals[use_index]

    if power <= 0 or signal <= 0:
        continue

    z_values_um.append(z_index * DEPTH_STEP_UM)
    log_signal_over_power_squared.append(np.log(signal / (power**2)))

z_arr = np.array(z_values_um)
y_arr = np.array(log_signal_over_power_squared)

# Since y = log(S/P²) decreases approximately linearly with depth,
# EAL is calculated from the fitted slope as EAL = -2 / slope.
eal_slope, eal_intercept = np.polyfit(z_arr, y_arr, 1)
y_pred = eal_slope * z_arr + eal_intercept
attenuation_length = -2 / eal_slope
r_squared = coefficient_of_determination(y_arr, y_pred)
```

```python
# Plot the linear EAL fit.
plt.figure(figsize=EAL_FIGSIZE)
plt.scatter(z_arr, y_arr, label="Data")

z_fit = np.linspace(z_arr.min(), z_arr.max(), 100)
y_fit = eal_slope * z_fit + eal_intercept
plt.plot(z_fit, y_fit, label=f"Fit: y={eal_slope:.6f}z + {eal_intercept:.2f}", linestyle="--")

plt.xlabel("Depth (µm)")
plt.ylabel("log(S / P²)")
plt.title("Effective Attenuation Length")
plt.text(
    0.05,
    0.05,
    f"EAL = {attenuation_length:.2f} µm\nR² = {r_squared:.4f}",
    transform=plt.gca().transAxes,
    verticalalignment="bottom",
)

plt.legend()
plt.tight_layout()
plt.savefig(EAL_PLOT_PATH, dpi=300, bbox_inches="tight")
plt.show()
```
